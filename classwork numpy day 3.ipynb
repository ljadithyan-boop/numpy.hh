```python
import numpy as np
arr=np.array([60,70,80,90])
x=np.array([65, 75, 85, 95])
print("the added sum is:",arr+x)
print("the difference id:",x-arr)
print("the increased percentage of term 1 :",arr*1.05)
print("the increased percentage of term 2:", x*1.05)
```

```python
import numpy as np
total_marks=(arr+x)
print("the total marks is:",total_marks)
print("the marks greater than 150 is:",total_marks>150)
#the marks between 140 and 170 is:
print((total_marks>140)&(total_marks>170))
result=total_marks.sum()
print("the sum is:",result)
result=total_marks.mean()
print("the mean is:",result)
result=total_marks.max()
print("the max mark is:",result)
result=total_marks.min()
print("the min mark is:",result)
result=total_marks.std()
print("the standard deviation is:",result)
```
