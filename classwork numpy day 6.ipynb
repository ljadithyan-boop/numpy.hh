```python
import numpy as np
arr=np.array([25.5, 26.0, np.nan, 27.8, np.inf, 26.4])
print("the original data is:",arr)
print("the detection of nan values:")
print(np.isnan(arr))
print("the detection of missing values:")
print(np.isinf(arr))
arr[np.isnan(arr)] = 0
arr[np.isinf(arr)] = 0
print(arr)

```
