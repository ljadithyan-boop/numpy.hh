```python
import numpy as np
arr=np.array([35000, 42000, np.nan, 50000, np.inf, 38000, np.nan])
print("the detection of nan values are:")
print(np.isnan(arr))
print("the detection of inf values are:")
print(np.isinf(arr))
arr[np.isnan(arr)] = 30000
arr[np.isinf(arr)] = 30000
print("the creation of new array are:",arr)
arr = np.array([35000, np.nan, 42000, np.inf, 70000])
clean_arr = arr[~np.isnan(arr)]
clean_arr = clean_arr[~np.isinf(clean_arr)]
print(clean_arr)
import numpy as np
arr = np.array([10, 20, 30,40])
copy_arr = arr.copy()
view_arr = arr.view()
view_arr[0] = 10
d=arr.view()
d[1]=999
print(arr)
#4.1 A view does not create a new array.
# It shares the same memory with the original array.
# Therefore, modifying the view also modifies the original array.
# 4.2 Since copy() creates new memory and stores the data separately,
# it uses more memory than a view.
```
