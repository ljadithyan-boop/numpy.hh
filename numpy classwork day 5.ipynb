```python
import numpy as np
arr=np.array([78, 85, 62, 90, 55, 88])
print("the sum of the array is :")
print(np.sum(arr))
print("the mean of the array is :")
print(np.mean(arr))
print("the median of the array is :")
print(np.median(arr))
print("the standard deviation of array is:")
print(np.std(arr))
print("the minimum marks of the array is:")
print(np.min(arr))
print("the maximum marks of the array is :")
print(np.max(arr))
grace_marks = arr + 5
bonus_marks = arr * 1.1
print("the grace marks of each student is:",grace_marks)
print("the bonus marks of each students is:",bonus_marks)
```
