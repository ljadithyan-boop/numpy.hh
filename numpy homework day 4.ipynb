```python
import numpy as np
arr=np.array([25, 45, 65, 85, 95])
result=np.where(arr<=40,"poor","good")
print("the results are:",result)
condition=[arr<=40,arr<=70,arr<=90]
choices=["poor","average","good"]
result=np.select(condition,choices,default="excellent")
print("the results using select are:",result)
arr=np.array([101, 102, 101, 103, 102, 104])
print(np.unique(arr))
print(np.any(arr>90))
print(np.all(arr>30))
```
