```python
import numpy as np
sales=np.array([250.75, 300.40, 275.60, 400.90, 150.30, 500.50, 350.80])
print("the total sales is :")
print(np.sum(sales))
print("the mean of the sales is :")
print(np.mean(sales))
print("the median of the sales is:")
print(np.median(sales))
print("the standard deviation is :")
print(np.std(sales))
print("the minimum value of sales:")
print(np.min(sales))
print("the maximum no of sales:")
print(np.max(sales))
print("the floor to sales is:")
print(np.floor(sales))
print("the ceil to sales is:")
print(np.ceil(sales))
print("the round to sales is:")
print(np.round(sales))
print("the exponential growth of sales is:")
print(np.exp(sales))
print("the cumulative sum of sales is:")
print(np.cumsum(sales))
a=np.array( [[2, 1],
[3, 4]])
b =np.array( [[1, 2],
[5, 6]])
print("the matrix multiplication is :")
print(a@b)
print(np.dot(a,b))
print("the determinant of a is :")
print(np.linalg.det(a))
print("the inverse of a:")
inv_a = np.linalg.inv(a)
print(inv_a)
values,vectors=np.linalg.eig(a)
print("the eigen values:",values)
print("the eigen vectors:",vectors)
a=np.array([[2,1],[3,4]])
b=np.array([[5,11]])
x=np.linalg.solve(a,b)
print("the linear equation solved:",x)
```
