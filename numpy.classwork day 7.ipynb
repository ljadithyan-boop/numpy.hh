```python
import numpy as np
import sys
celsius = np.array([28, 30, 32, 35, 33])
fahrenheit_loop = []
for temp in celsius:
    fahrenheit_loop.append((temp * 9/5) + 32)
print("Method 1: Fahrenheit using Python for loop")
print(fahrenheit_loop)
fahrenheit_numpy = (celsius * 9/5) + 32
print("Method 2: Fahrenheit using NumPy Vectorized Operation")
print(fahrenheit_numpy)
python_list = [28, 30, 32, 35, 33]
numpy_array = np.array([28, 30, 32, 35, 33])
print("Memory Usage")
print("Python List:", sys.getsizeof(python_list), "bytes")
print("NumPy Array:", numpy_array.nbytes, "bytes")
np.save("fahrenheit.npy", fahrenheit_numpy)
print("Vectorized Fahrenheit array saved to 'fahrenheit.npy'")
loaded_array = np.load("fahrenheit.npy")
print("Loaded Fahrenheit Array")
print(loaded_array)
```
